```python
# 5_1 С помощью функций получить вертикальную и горизонтальную линии.
# Линия проводится многократной печатью символа. Заключить слово в рамку
# из полученных линий.


def horizontal_line(char, length):
    print(char * length)


def vertical_line(char, word):
    print(char + word + char)


word = input("Введите слово: ")
width = len(word) + 2  # ширина рамки: слово плюс по символу с боков
horizontal_line("*", width)  # верхняя линия
vertical_line("*", word)     # вертикальные линии со словом
horizontal_line("*", width)  # нижняя линия
```

```python
# 5_2 Описать функцию ShiftLeft3(A, B, C), выполняющую левый
# циклический сдвиг: значение А переходит в С, значение С — в В,
# значение В — в А. С помощью этой функции выполнить левый циклический
# сдвиг для двух наборов из трех чисел.


def shift_left3(a, b, c):
    return b, c, a  # B -> A, C -> B, A -> C


for n in (1, 2):
    while True:  # обработка исключений
        try:
            a = float(input(f"Введите число A{n}: "))
            b = float(input(f"Введите число B{n}: "))
            c = float(input(f"Введите число C{n}: "))
            break
        except ValueError:
            print("Неправильно ввели!")
    a, b, c = shift_left3(a, b, c)
    print("После сдвига: A =", a, "B =", b, "C =", c)
```
