```python
# 7_1 Дана строка. Подсчитать количество содержащихся в ней
# прописных латинских букв.

s = input('Введи строку: ')
print('Исходная строка:', s)

count = 0  # счетчик прописных латинских букв
for char in s:
    if 'A' <= char <= 'Z':  # проверка диапазона прописных латинских букв
        count += 1

print('Результирующая строка:', s)
print('Количество прописных латинских букв:', count)
```

```python
# 7_2 Дана строка-предложение на русском языке. Подсчитать
# количество содержащихся в строке гласных букв.

VOWELS = 'аеёиоуыэюяАЕЁИОУЫЭЮЯ'  # гласные буквы русского алфавита

s = input('Введи предложение: ')
print('Исходная строка:', s)

count = 0  # счетчик гласных букв
for char in s:
    if char in VOWELS:  # проверка принадлежности гласным
        count += 1

print('Результирующая строка:', s)
print('Количество гласных букв:', count)
```
