```python
#3_1 Даны три целых числа: A, B, C. Проверить истинность высказывания: «Каждое из
#чисел A, B, C положительное»
a = int(input("Введите A: "))
while type(a) != int: #обработка исключений
    try:
        a = int(a)
    except ValueError:
        print("Неправельный ввод!")
        a = input("Введите A: ")

b = int(input("Введите B: "))
while type(a) != int: #обработка исключений
    try:
        b = int(b)
    except ValueError:
        print("Неправельный ввод!")
        b = input("Введите b: ")

c = int(input("Введите C: "))
while type(a) != int: #обработка исключений
    try:
        c = int(c)
    except ValueError:
        print("Неправельный ввод!")
        c = input("Введите C: ")

if a > 0 and b > 0 and c > 0:
     print("Каждое из чисел положительно")
else:
     print("Каждое из чисел отрицательно")
```

```python
#3_2 Даны три числа. Найти среднее из них (то есть число, расположенное между
#наименьшим и наибольшим)
while True:
  try:
      a = float(input("Введите первое число: "))
      break
  except ValueError:
      print("Ошибка! Введите число")

while True:
  try:
      b = float(input("Введите первое число: "))
      break
  except ValueError:
      print("Ошибка! Введите число")

while True:
  try:
      c = float(input("Введите первое число: "))
      break
  except ValueError:
      print("Ошибка! Введите число")

nums = [a, b, c]
nums.sort()
middle = nums[1]  # После сортировки это и есть среднее по величине

print("Среднее число:", middle)
```
