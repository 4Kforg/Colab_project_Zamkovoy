```python
#3_1 Даны три целых числа: A, B, C. Проверить истинность высказывания: «Каждое из
#чисел A, B, C положительное»
a = int(input("Введите A: "))
while type(a) != int: #обработка исключений
    try:
        a = int(a)
    except ValueError:
        print("Неправельный ввод!")
        a = input("Введите A: ")

b = int(input("Введите B: "))
while type(a) != int: #обработка исключений
    try:
        b = int(b)
    except ValueError:
        print("Неправельный ввод!")
        b = input("Введите b: ")

c = int(input("Введите C: "))
while type(a) != int: #обработка исключений
    try:
        c = int(c)
    except ValueError:
        print("Неправельный ввод!")
        c = input("Введите C: ")

if a > 0 and b > 0 and c > 0:
     print("Каждое из чисел положительно")
else:
     print("Каждое из чисел отрицательно")
```

```python
# 3_2 Даны три числа.
# Найти среднее из них (то есть число, расположенное между наименьшим и наибольшим).


def read_float(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Неправильно ввели!")


def find_middle(a, b, c):
    if (b <= a <= c) or (c <= a <= b):
        return a
    if (a <= b <= c) or (c <= b <= a):
        return b
    return c


def main():
    a = read_float("Введите первое число: ")
    b = read_float("Введите второе число: ")
    c = read_float("Введите третье число: ")

    middle = find_middle(a, b, c)
    print(f"Среднее из чисел {a}, {b}, {c} — это {middle}")


if __name__ == "__main__":
    main()
```
