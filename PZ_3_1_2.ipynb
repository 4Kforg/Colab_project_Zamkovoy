```python
#3_1 Даны три целых числа: A, B, C. Проверить истинность высказывания: «Каждое из
#чисел A, B, C положительное»
A = int(input("Введите A: "))
B = int(input("Введите B: "))
C = int(input("Введите C: "))

result = A > 0 and B > 0 and C > 0
print(result)  # Выведет True или False
```

```python
#3_2 Даны три числа. Найти среднее из них (то есть число, расположенное между
#наименьшим и наибольшим)
a = float(input("Введите первое число: "))
b = float(input("Введите второе число: "))
c = float(input("Введите третье число: "))

nums = [a, b, c]
nums.sort()
middle = nums[1]  # После сортировки это и есть среднее по величине

print("Среднее число:", middle)
```
