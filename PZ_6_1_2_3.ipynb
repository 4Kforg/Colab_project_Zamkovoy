```python
# 6_1 Дан список A размера N. Вывести вначале его элементы с нечетными
# номерами в порядке возрастания, а затем — с четными номерами в порядке
# убывания: A1, A3, A5, ..., A6, A4, A2. Условный оператор не использовать.

import random

n = int(input('Введи размер списка: '))
a = [random.randint(1, 100) for _ in range(n)]  # заполнение списка случайными числами

print('Исходный список:', a)
# a[::2] — нечетные номера, a[::-2] — четные номера в порядке убывания
print('Результат:', a[::2] + a[::-2])
```

```python
# 6_2 Дан список A размера N. Сформировать два новых списка B и C:
# в список B записать все положительные элементы списка A, в список C —
# все отрицательные (сохраняя исходный порядок). Вывести размер и содержимое
# списка B, а затем — размер и содержимое списка C.

import random

n = int(input('Введи размер списка: '))
a = [random.randint(-50, 50) for _ in range(n)]

print('Исходный список:', a)
b = [x for x in a if x > 0]  # положительные элементы
c = [x for x in a if x < 0]  # отрицательные элементы
print('Список B. Размер:', len(b))
print(b)
print('Список C. Размер:', len(c))
print(c)
```

```python
# 6_3 Дан список размера N. Осуществить циклический сдвиг элементов
# списка вправо на одну позицию (A1 перейдет в A2, A2 — в A3, ..., AN — в A1).

import random

n = int(input('Введи размер списка: '))
a = [random.randint(1, 100) for _ in range(n)]

print('Исходный список:', a)
a = [a[-1]] + a[:-1]  # последний элемент становится первым
print('Результат после сдвига:', a)
```
