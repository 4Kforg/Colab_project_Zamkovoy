```python
#4_1 Дано вещественное число A и целое число N (>0). Используя один цикл, вывести все
#целые степени числа A от 1 до N
A = float(input("Введите вещественное число A: "))
N = int(input("Введите целое число N (> 0): "))

power = A  # A^1
for i in range(1, N + 1):
    print(f"A^{i} = {power}")
    power *= A  # Готовим A^(i+1) для следующей итерации
```

```python
#4_2 Дано целое число N (>0). Найти наименьшее целое положительное число K, квадрат
#которого превосходит N: K2 > N. Функцию извлечения квадратного корня не
#использовать
N = int(input("Введите целое число N (> 0): "))

K = 1
while K * K <= N:
    K += 1

print("Наименьшее K, такое что K^2 > N:", K)
```
